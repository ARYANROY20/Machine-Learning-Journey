```python
#Importing Pandas
import pandas as pd
```

```python
#Creating a Series from a List
data=[1,2,3,4,5]
series=pd.Series(data)
print(series)
print(type(series))
```

```python
#Creating a Series from a Dictionary
data={
    'a':1,
    'b':2,
    'c':3
}
series=pd.Series(data)
print(series)
```

```python
#Creating a Series with a Custom Index
data=[10,20,30]
index=['a','b','c']
pd.Series(data,index=index)
```

```python
#Creating a DataFrame from a Dictionary
data={
    'Name':['Aryan','Krish','John'],
    'Age':[20,30,45],
    'City':['Kolkata','Bangalore','Florida'] 
}
df=pd.DataFrame(data)
df
```
