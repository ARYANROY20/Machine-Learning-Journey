```python
# Reference counting example
import sys
a=[]
print(sys.getrefcount(a))
```

```python
# Creating a shared reference
b=a
print(sys.getrefcount(a))
```

```python
# Removing a reference
del b
print(sys.getrefcount(a))
```

```python
# Enabling garbage collection
import gc
gc.enable()
gc.collect()
```

```python
# Checking GC statistics
print(gc.get_stats())
```

```python
# Inspecting uncollectable garbage
print(gc.garbage)
```

```python
# Testing circular references
import gc 
class MyObject:
    def __init__(self,name):
        self.name=name
        print(f"Object {self.name} created successfully")
    def __del__(self):
        print(f"object {self.name} deleted")


obj1=MyObject("Obj1")
obj2=MyObject("Obj2")
obj1.ref=obj2
obj2.ref=obj1

gc.collect()
print(f"Garbage collected objects:{gc.garbage}")
```

```python
#Generators for Memory Efficiency
def generate_numbers(n):
    for i in range(n):
        yield i

for num in generate_numbers(1000):
    print(num)
    if num>10:
        break
```

```python
#Profiling Memory Usage with tracemalloc
import tracemalloc

def create_list():
    return [i for i in range(1000)]

def main():
    tracemalloc.start()

    my_list = create_list()

    snapshot = tracemalloc.take_snapshot()
    top_stats = snapshot.statistics("lineno")

    print("[Top 10]")
    for stats in top_stats[:10]:
        print(stats)

    tracemalloc.stop()

main()
```
