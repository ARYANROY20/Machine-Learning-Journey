```python
def fibonacci(n,memo={}):
    if n in memo:
        return memo[n]
    if n<=1:
        return n
    memo[n]=fibonacci(n-1,memo)+fibonacci(n-2,memo)
    return memo[n]

print(fibonacci(10))  
print(fibonacci(15))
```

```python
def func(a,b={}):
    b[a]=a**2
    return b

print(func(2)) 
print(func(3, {1: 1})) 
```

```python
def filter_integers(**kwargs):
    dict_int={}
    for key,value in kwargs.items():
        if isinstance(value, int):
            dict_int[key]=value
    return dict_int



print(filter_integers(a=1, b='two', c=3, d=4.5))  # {'a': 1, 'c': 3}
print(filter_integers(x=10, y='yes', z=20))  # {'x': 10, 'z': 20}
```

```python
def apply_callback(callback,lst):
    return [callback(x) for x in lst]

print(apply_callback(lambda x:x**2,[1, 2, 3, 4]))
```

```python
def outer_function():
    def inner_function(x):
        return x**2
    return inner_function

square=outer_function()
print(square(2))
```

```python
def filter_map(filter_func,map_func,lst):
    return [map_func(x) for x in lst if filter_func(x)]

print(filter_map(lambda x:x%2==0,lambda x:x**2,[1,2,3,4,5]))
```

```python
def compose(f, g):
    return lambda x: f(g(x))


f = lambda x: x + 1
g = lambda x: x * 2
h = compose(f, g)
print(h(3))  
print(h(5))  
```

```python
from functools import partial #It lets you fix some arguments of another function in advance

multiply_by_2 = partial(lambda x, y: x * y, 2)


print(multiply_by_2(3))  
print(multiply_by_2(5)) 
```

```python
def average(lst):
    try:
        return sum(lst)/len(lst)
    except ZeroDivisionError:
        return None
print(average([1, 2, 3, 4, 5])) 
print(average([])) 
```

```python
#A curried function is a function that takes multiple arguments one at a time,
#using a sequence of functions, where each function returns another function until all arguments are provided.

def curry_product(x):
    def inner1(y):
        def inner2(z):
            return x*y*z
        return inner2
    return inner1

print(curry_product(2)(3)(4)) 
```

```python
def separate_types(lst):
    ints, strs, floats = [], [], []
    for item in lst:
        if isinstance(item, int):
            ints.append(item)
        elif isinstance(item, str):
            strs.append(item)
        elif isinstance(item, float):
            floats.append(item)
    return ints, strs, floats

print(separate_types([1, 'a', 2.5, 3, 'b', 4.0, 'c']))
```

```python
def call_counter(counter={'count': 0}):
    counter['count'] += 1
    return counter['count']

print(call_counter()) 
print(call_counter())  
print(call_counter()) 
```
